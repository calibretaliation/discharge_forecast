```python
import numpy as np
import pandas as pd

# Create a NumPy array
data = np.array([[10, 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]])

# Convert to DataFrame with default integer index and column names
df = pd.DataFrame(data)
print("DataFrame with default settings:")
print(df)
print("DataFrame Overview:")
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())
print("\nStatistical Summary:")
print(df.describe())
```
